```python
import numpy as np 
from matplotlib import pyplot as plt
import os
from os import path
import h5py
import pprint as pp
```

```python
speed_of_light = 2.99792458e10
stefan_boltzman = 5.670374e-5
radiation_constant = 4 * stefan_boltzman / speed_of_light
boltzmann_constant = 1.380649e-16
ev = 1.602176634e-12
kev = 1e3*ev

kev_to_kelvin = kev / boltzmann_constant

dir_figs = 'densmore'
output_dir = '/home/itamarg/workspace/RICH/output/'
data_dir = '/home/itamarg/workspace/RICH/densmore_data'

os.makedirs(dir_figs, exist_ok=True)
```

```python
output_files_path = "/home/itamarg/workspace/RICH/output/densmore2012_10kev"
path_to_final_state = path.join(output_files_path, f"final_state.h5")

num_cells = None
num_groups = None

with h5py.File(path_to_final_state) as h5_file:
    print(h5_file.keys())
    num_cells = len(np.array(h5_file["Temperature"]))
    num_groups = len([key for key in h5_file.keys() if "Eg" in key])

print("num_cells: ", num_cells)
print("num_groups: ", num_groups)

Tmat = np.ndarray(shape=num_cells)
time = None
CMx = np.ndarray(shape=num_cells)
with h5py.File(path_to_final_state) as h5_file:
    Tmat = np.array(h5_file["Temperature"])/kev_to_kelvin
    CMx = np.array(h5_file["CMx"])
    time = h5_file["Time"][0]


plt.plot(CMx, Tmat, label="RICH")
path_to_data = path.join(data_dir, "10kev.txt")
data = np.loadtxt(path_to_data)
plt.plot(data[::3, 0], data[::3, 1], 'o', c='k', label="Densmore")
plt.xlim(0., 5.)
plt.ylim(0., 0.9)
plt.grid(True)
plt.xlabel(r"$x\ [cm]$")
plt.ylabel(r"$T_{mat}\ [KeV]$")
plt.title(f"Time = {time/1e-9:.2f} [ns]")
plt.legend(loc="best")
plt.savefig(path.join(dir_figs, "densmore2012_10kev.png"))
plt.show()
plt.close()
```

```python
output_files_path = "/home/itamarg/workspace/RICH/output/densmore2012_step_2"
path_to_final_state = path.join(output_files_path, f"final_state.h5")

num_cells = None
num_groups = None

with_mpi = True

Tmat = None
time = None
CMx = None

if not with_mpi:
    with h5py.File(path_to_final_state) as h5_file:
        print(h5_file.keys())
        num_cells = len(np.array(h5_file["Temperature"]))
        num_groups = len([key for key in h5_file.keys() if "Eg" in key])

    print("num_cells: ", num_cells)
    print("num_groups: ", num_groups)

    with h5py.File(path_to_final_state) as h5_file:
        Tmat = np.array(h5_file["Temperature"])/kev_to_kelvin
        CMx = np.array(h5_file["CMx"])
        time = h5_file["Time"][0]
else:
    with h5py.File(path_to_final_state) as h5_file:
        print(h5_file.keys())
        num_ranks = len([key for key in h5_file.keys() if "rank" in key])
        num_groups = len([key for key in h5_file['rank0'].keys() if "Eg" in key])
        
        num_cells = 0
        for i in range(num_ranks):
            num_cells += h5_file[f"rank{i}"]['Temperature'].shape[0]

        print("num_ranks: ", num_ranks)    
        print("num_cells: ", num_cells)
        print("num_groups: ", num_groups)

        time = h5_file['Time'][0]
        Tmat = np.ndarray(shape=(0))
        CMx =  np.ndarray(shape=(0))
        for i in range(num_ranks):
            num_cells += h5_file[f"rank{i}"]['Temperature'].shape[0]
            Tmat = np.append(Tmat, np.array(h5_file[f"rank{i}"]['Temperature'])/kev_to_kelvin)
            CMx = np.append(CMx, np.array(h5_file[f"rank{i}"]['CMx']))








plt.plot(CMx, Tmat, label="RICH")
path_to_data = path.join(data_dir, "step_2.txt")
data = np.loadtxt(path_to_data, delimiter=',')
plt.plot(data[::2, 0], data[::2, 1], 'o', c='k', label="Densmore")
plt.xlim(0., 1.5)
# plt.ylim(0.2, 1.0)
plt.grid(True)
plt.xlabel(r"$x\ [cm]$")
plt.ylabel(r"$T_{mat}\ [KeV]$")
plt.title(f"Time = {time/1e-9:.2f} [ns]")
plt.legend(loc="best")
plt.savefig(path.join(dir_figs, f"densmore2012_step_2.png"))
plt.show()
plt.close()
```
